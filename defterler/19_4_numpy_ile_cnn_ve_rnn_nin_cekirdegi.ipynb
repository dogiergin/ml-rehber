Derin öğrenme kütüphaneleri bu işlemleri çok daha hızlı yapar; ama mantığı görmek için birkaç satır yeterlidir. Aşağıdaki kod Bölüm 15'teki evrişim örneğini ve Bölüm 16'daki RNN adımını hesaplar:

```python
import numpy as np

def evrisim2d(goruntu, filtre):
    """Dolgu yok (padding=0), adım 1 (stride=1) ile 2B evrişim."""
    H, W = goruntu.shape
    k = filtre.shape[0]
    cikti = np.zeros((H - k + 1, W - k + 1))
    for i in range(cikti.shape[0]):
        for j in range(cikti.shape[1]):
            pencere = goruntu[i:i + k, j:j + k]
            cikti[i, j] = np.sum(pencere * filtre)   # çarp ve topla
    return cikti

goruntu = np.array([[0, 0, 0, 9, 9]] * 5)   # solu koyu, sağı açık
dikey_kenar = np.array([[-1, 0, 1]] * 3)    # dikey kenar filtresi
print(evrisim2d(goruntu, dikey_kenar))      # her satır: [ 0. 27. 27.]

def rnn_adimi(x_t, h_onceki, W_x, W_h, b):
    """Basit RNN: yeni gizli durum = tanh(W_x·x_t + W_h·h_önceki + b)"""
    return np.tanh(W_x @ x_t + W_h @ h_onceki + b)

rng = np.random.default_rng(0)
W_x, W_h, b = rng.normal(size=(4, 3)), rng.normal(size=(4, 4)), np.zeros(4)
h = np.zeros(4)                          # başlangıç hafızası
for x_t in rng.normal(size=(5, 3)):      # 5 adımlık bir dizi
    h = rnn_adimi(x_t, h, W_x, W_h, b)   # aynı ağırlıklar her adımda
print(h.shape)                           # (4,)
```

İlk çıktı, Şekil 15.1'deki özellik haritasıdır (her satır 0, 27, 27). PyTorch'ta aynı yapı taşları hazır katmanlar olarak bulunur: `nn.Conv2d` (evrişim), `nn.MaxPool2d` (havuzlama), `nn.LSTM` ve `nn.GRU` (tekrarlayan katmanlar), `nn.TransformerEncoderLayer` (Transformer bloğu). Keras'taki karşılıkları `Conv2D`, `MaxPooling2D`, `LSTM`, `GRU` ve `MultiHeadAttention`'dır. Bu katmanlarla kurulmuş, baştan sona çalışan iki örnek Bölüm 20'de.
